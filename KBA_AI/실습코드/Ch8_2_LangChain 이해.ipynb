### LCEL(LangChain Expression Language)이해

```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser

prompt = PromptTemplate.from_template("{topic}에 대해 쉽게 설명해주세요.") # 프롬프트 템플릿
model = ChatOpenAI() # LLM 모델 default "gpt-3.5-turbo"
output_parser = StrOutputParser() # 출력 파서
chain = prompt | model | output_parser  # LCEL을 이용하여 체인 구성
result = chain.invoke({"topic": "인공지능"})   # 체인 실행
print(result)   # 결과 출력
```

### LLM 모델
▶ llm 객체를 생성하며 옵션 설정

```python
from langchain_openai import ChatOpenAI
# 객체 생성
llm = ChatOpenAI(
    temperature=0.1,	# 창의성 (0.0 ~ 2.0)
    model="gpt-4o-mini",	# 모델명
    max_tokens= 100,
    # api_key= "sk- 실습시 발급한 키 입력"
)

llm.invoke("하늘은 왜 파랗나요?") 
# llm.invoke("하늘은 왜 파랗나요?").content  # content만 출력
```

```python
llm.invoke("하늘은 왜 파랗나요?").content  # content만 출력
```

### Parser
▶ llm의 출력을 구조화된 형태로 변환하는 출력파서(Output Parser)

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. LLM 생성
llm = ChatOpenAI(
    model="gpt-4o-mini",
    # api_key="sk-실습시 발급한 키 입력"
)

# 2. LLM에게 질문
response = llm.invoke("2차전지에 대해 한 문장으로 쉽게 설명해줘")

# 3. LLM의 원본 응답 확인
print(response)
```

```python
# 4. Output Parser 생성
parser = StrOutputParser()

# 5. 응답 객체에서 실제 답변 텍스트만 추출
result = parser.invoke(response)

# 6. 최종 결과 출력
print(result)
```

### 영어회화 출력하기

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

# 프롬프트 템플릿을 이용하여 프롬프트 생성 
# 객체 생성
template = """
당신은 영어 선생님입니다. 요청에 맞춰 [format]에 영어 회화를 작성해 주세요.

요청:
{question}

format:
- 영어 회화:
- 한글 해석:
"""

# 템플릿 생성
prompt = PromptTemplate.from_template(template)	

model = ChatOpenAI(
     model_name="gpt-4o", 
     # api_key= "sk- 실습시 발급한 키 입력"
     )

# OutputParser
output_parser = StrOutputParser()	# 문자열 출력 파서를 초기화

chain = prompt | model | output_parser	# 체인 구성
scripts = chain.invoke({"question": "비행기표를 예매하고 싶어요"})	# 출력

print(scripts)
```
