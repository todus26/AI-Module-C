# **8장 Langchain이해**

### LCEL(LangChain Expression Language)이해

```python
#langchain 계열의 패키지들. 내 PC에 깔려있는 패키지로부터 함수를 가져옴
from langchain_core.prompts import PromptTemplate          # langchain core라는 패키지(폴더)의 prompts모듈(.py)로부터 promptTemplate 함수/기능(코드)를 가져옴
from langchain_openai import ChatOpenAI                
from langchain_core.output_parsers import StrOutputParser  # 출력에 대한 것. Key, Value 형태의 딕셔너리 형태로 데이터를 저장하고 관리하는 자료구조

prompt = PromptTemplate.from_template("{topic}에 대해 쉽게 설명해주세요.") # 프롬프트 템플릿. from_template: 메서드, PromptTemplate: 클래스
# 데이터와 함수(기능)를 묶어서 객체를 만듦. 클래스 안에 함수가 있고 그를 갖다씀. 
# {}는 변수를 받기 위함. 

model = ChatOpenAI()                                   # model=객체.                # LLM 모델 default "gpt-3.5-turbo"
output_parser = StrOutputParser() # 출력 파서
chain = prompt | model | output_parser  # LCEL을 이용하여 체인 구성
result = chain.invoke({"topic": "인공지능"})   # 체인 실행.
# invoke : 한 번 실행하는 것. stream : 한 줄씩 받아오는 것. 

print(result)   # 결과 출력
```

### LLM 모델
▶ llm 객체를 생성하며 옵션 설정
- 정확하게 흔들리지 않게, 규칙에 맞게, AI가 더 잘 이해할 수 있도록 의도와 목적을 구조화시키기 위해 이렇게 표현. 

```python
from langchain_openai import ChatOpenAI   # 모델 설정 
# 객체 생성
llm = ChatOpenAI(               # 이 모듈의 인자들. 
    temperature=0.1,	        # 창의성 (0.0 ~ 2.0) - 0.1 : 사실에 근거해 추측하지 말고 답변 
    model="gpt-4o-mini",	    # 모델명 - 이걸 안 쓰면 3.5-turbo 
    max_tokens= 100,            # 최대 토큰 수 - 100 : 100개의 토큰만 생성
    # api_key= "sk- 실습시 발급한 키 입력" - 우리는 환경변수로 사용. 
    # api_key는 하드코딩 안 하는게 좋음. 
)

llm.invoke("하늘은 왜 파랗나요?") 
# llm.invoke("하늘은 왜 파랗나요?").content  # content만 출력
```

```python
llm.invoke("하늘은 왜 파랗나요?").content  # content만 출력
```

### Parser
▶ llm의 출력을 구조화된 형태로 변환하는 출력파서(Output Parser)
- 출력물을 구조화시키는 출력물 파서. 

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. LLM 생성
llm = ChatOpenAI(
    model="gpt-4o-mini",
    # api_key="sk-실습시 발급한 키 입력"
)

# 2. LLM에게 질문
response = llm.invoke("2차전지에 대해 한 문장으로 쉽게 설명해줘")

# 3. LLM의 원본 응답 확인
print(response)     
```

```python
# 4. Output Parser 생성
# 문장이 길어지면, 처리하는게 불편해짐. 그래서 다음 프로그램과의 연결성을 강화시켜주고, 다음 프로그램이 content를 갖다쓰기 편하게 하기 위해서 함수 활용
parser = StrOutputParser()

# 5. 응답 객체에서 실제 답변 텍스트만 추출
result = parser.invoke(response)

# 6. 최종 결과 출력
print(result)
```

### 영어회화 출력하기

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import PromptTemplate

# 프롬프트 템플릿을 이용하여 프롬프트 생성 
# 객체 생성
template = """
당신은 영어 선생님입니다. 요청에 맞춰 [format]에 영어 회화를 작성해 주세요.

요청:
{question}

format:
- 영어 회화:
- 한글 해석:
"""

# 템플릿 생성
prompt = PromptTemplate.from_template(template)	

model = ChatOpenAI(
     model_name="gpt-4o", 
     # api_key= "sk- 실습시 발급한 키 입력"
     )

# OutputParser
output_parser = StrOutputParser()	# 문자열 출력 파서를 초기화

chain = prompt | model | output_parser	# 체인 구성
scripts = chain.invoke({"question": "비행기표를 예매하고 싶어요"})	# 출력

print(scripts)
```
