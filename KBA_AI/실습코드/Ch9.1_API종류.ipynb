### Chat Completions
- 에이전트의 기본 뇌 역할. 무상태성(Stateless) API로 매번 과거 대화를 직접 묶어서 보내야 함.

한번 질문 하면 땡. 연결(연계)을 못 함.

Openai에서 제공하는 여러 가지 api들. 

```python
import os
import requests
url = "https://api.openai.com/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}",  # Bearer : ~를 갖고 있는. 즉, API Key를 갖고 있는.
    "Content-Type": "application/json",}                        # 안내에 대한 것. API Key를 여기서 받음. 

data = {    "model": "gpt-4o-mini",
    "messages": [{"role": "user", "content": "안녕? 내 이름은 유익상이야"}],}

# 서버로 POST 요청 전송
response = requests.post(url, headers=headers, json=data)
print(response.json())

# 얘는 기억을 못 함. 
```

앱과 DB가 API로 통신하는 방법.
- GET : 가져와라
- POST : 주문하는 것
- PUT : 전체 수정하는 것. 
- PATCH : 일부만 수정 
- DELETE : 삭제 

```python
url = "https://api.openai.com/v1/chat/completions"

headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}", 
    "Content-Type": "application/json",}

data = {    "model": "gpt-4o-mini",
    "messages": [{"role": "user", "content": "내 이름이 뭐라고 했지?"}],}

# 서버로 POST 요청 전송
response = requests.post(url, headers=headers, json=data)
print(response.json())

# 기억을 못 해!!!
```

###  Responses API
- 상태유지형(Stateful) API. thread_id로 서버가 대화 맥락을 기억함. 비용 절감 및 관리 용이.
- 전에 내가 보낸 정보를 기억해줌. 연결을 시켜줘야 함. 

```python
import os
import requests

url = "https://api.openai.com/v1/responses"
headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}",
    "Content-Type": "application/json",
}

# 첫번째 질문
data = {
    "model": "gpt-4o-mini",
    "input": "안녕? 내 이름은 유익상이야"
}

response = requests.post(url, headers=headers, json=data)
result = response.json()

# 첫 번째 응답의 ID 저장
previous_id = result["id"]
print(result["output"][0]["content"][0]["text"])
```

```python
# 두 번째 질문
data = {
    "model": "gpt-4o-mini",
    "input": "내 이름이 뭐라고 했지?",
    "previous_response_id": previous_id  # 이전 셀에 존재. 이렇게 연결시켜놓은 것. 
}

response = requests.post(url, headers=headers, json=data)
result = response.json()
print(result["output"][0]["content"][0]["text"])
```

### Image 생성 API: v1/images

#### 1) Image 생성: DALL-E(이미지 생성)등의 모델 연결.

- 내부적으로 연결해놔서 답변하는 거임. 

```python
import os
import requests
import base64

url = "https://api.openai.com/v1/images/generations"

headers = {
    "Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}",
    "Content-Type": "application/json"
}

data = {
    "model": "gpt-image-2",
    "prompt": "LG엔솔 공장에서 AI가 품질검사를 하는 모습"
}

response = requests.post(
    url,
    headers=headers,
    json=data
)

result = response.json()
image_data = result["data"][0]["b64_json"]

with open("LG엔솔_ai.png", "wb") as f:
    f.write(base64.b64decode(image_data))

print("이미지 저장 완료")
```

#### 2)Audio API: v1/audio
- Whisper(음성 인식/STT)등의 모델 연결

```python
import os
import requests
url = "https://api.openai.com/v1/audio/transcriptions"

headers = {"Authorization": f"Bearer {os.environ['OPENAI_API_KEY']}"}
files = {"file": open("musk.m4a", "rb")}
data = {"model": "gpt-4o-transcribe"}
response = requests.post(
    url,
    headers=headers,
    files=files,
    data=data
)
english_text = response.json()['text']
print(english_text)
```

```python
from openai import OpenAI

client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": f"다음 영어 문장을 자연스러운 한국어로 번역해줘:\n\n{english_text}"}])
korean_text = response.choices[0].message.content
print(korean_text)
```
