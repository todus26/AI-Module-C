# 3) 조건문 ⭐ - 총 4가지 유형

### if문

```python
# 비교 연산자와 논리 연산자 활용
# 조건문 뒤에 콜론(:)을 붙이고, 실행할 코드는 반드시 들여쓰기(권장 표준 4칸)를 한다. 
# 비교 연산자와 논리 연산자(and, or, not) 활용

age = 25 
has_ticket = True
if age >= 20 and has_ticket: 
     print("입장이 가능합니다.")
```

```python
# 양자 택일 : if ~ else 문

money = 1000

if money >= 3000:           # 3000원 이상이면 택시를 탄다. 
    print("택시를 탄다")
else:                       # 3000원 미만이면 걸어간다.
     print("걸어간다")       # 조건이 거짓일 때 실행
```

```python
# 다중 선택 : if ~ elif ~ else - 점수 부여에 많이 사용. 불량에 대한 등급 판정 등. 데이터 분석에서 분류 문제 
# 여러 개 중 한 가지. 여러 조건을 부여할 수 있음. 

score = 85           # score 클릭하면 동일한 변수 선택됨. 
if score >= 90:    
       print("A학점")
elif  score >= 80:   
       print("B학점")  # 여기서 참이 되므로 실행하고 if문 종료
elif score >= 70:   
       print("C학점")
else:    
       print("F학점")
```

```python
# 중첩 조건문 - ⭐⭐들여쓰기가 맞아야 한다. 조건문 안에 조건문! 어떻게 보면 AND 조건인 것임. 
age = 20
card = True
if age >= 19:   
    if card:   
         print("결제 진행")    
    else:        
         print("카드가 필요합니다")
else:   
    print("미성년자는 구매 불가")
```

```python
for i in range(1, 4): # i에 1, 2, 3이 차례로 들어갑니다.
    print(f"{i}번째 반복 중...")  # 여기도 i로 해줘야 들어가는 거임. 
```

```python
# AI 모델을 데이터로 학습시킬 때, 반복 학습 횟수를 'Epoch(에폭)'이락고 하며, for문으로 구성된다. 
# 전체 데이터를 5번 반복해서 학습한다고 가정

epochs = 5
for epoch in range(1, epochs + 1):  #  1부터 시작하니, 5번해주기 위해 +1을 해준다. 
     print(f" [Epoch {epoch}/{epochs}] 학습을 시작합니다...")    # (여기서 실제로 복잡한 수학 연산과 학습이 일어납니다)    
     print(f" {epoch}번째 학습 완료! 오차(Error)가 줄어들었습니다.")    
     print("-" * 30)
print("모든 AI 모델 학습이 완료되었습니다!")
```

```python
epochs = 5
for epoch in range(1, epochs+1):
    print(f" [Epoch {epoch}/{epochs}] 학습을 시작합니다...")
    print(f" {epoch}번째 학습 완료! 오차(Error)가 줄어들었습니다.")
    print("-" * 30)
print("모든 AI 모델 학습이 완료되었습니다!")
```

### 리스트 컴프리헨션 
- for 반복문을 한 줄로 압축하여 새로운 리스트를 만든다.

```python
# 기본형: 표현식 for 항목 in 리스트 - 예) 모든 숫자를 10배로 만들기
# n*10 = 표현식, 
nums = [1, 2, 3, 4, 5]

result = [n * 10 for n in nums]         # 각각의 값에 10을 곱해, 새로운 리스트. 
print(result)   # [10, 20, 30, 40, 50]
```

```python
nums = [1,2,3,4,5]

result = [n**10 for n in nums]
print(result)
```

```python
# 필터링형: 표현식 for 항목 in 리스트  if 조건 - 예) 짝수만 남기기
# 기본형 뒤에 if 문이 붙는다. ~한다면으로 필터링해주는 거. 
# for문이 앞에!! if문이 뒤!!!

nums = [1, 2, 3, 4, 5]
evens = [n for n in nums if n % 2 == 0]
print(evens) # [2, 4]
```

```python
# 선택형: 참값  if 조건 else 거짓값 for 항목 in 리스트 - 예) 60점 이상이면 '합격', 아니면 '불합격’
scores = [80, 50, 90, 40]
results = ["합격" if s >= 60 else "불합격" for s in scores ]
print(results) 
```

### 가변 인자 (*args, **kwargs)

- arg : argument
- *하나 붙은 것 : 여러 변수를 순서대로 하나씩 대입시켜줌. 
    - 순서대로 값을 전달시키는 역할. 여러 값을 한 번에 전달.
- ** : 딕셔너리 형태로 받는다. key와 value의 형태

```python
# 위치 가변인자 (*args)
def sum_all(*args):
    # 여러 개의 인자를 튜플로 묶음
    return sum(args)

print(sum_all(10, 20))
print(sum_all(1, 2, 3, 4, 5))
```

```python
# 키워드 가변인자 (**kwargs)   ※ kwargs: Keyword Arguments의 약자
def print_user_info(**kwargs): 
# kwargs는 {'name': 'Gemini', 'age': 20} 처럼 딕셔너리로 묶입니다.
   for key, value in kwargs.items():
     print(f"{key}: {value}") 

print_user_info(name="Gemini", version="3.0")
```

```python
def print_user(**kw):
    for k, v in kw.items():
        print(f"{k}: {v}")

print_user(name="GPT", version="2.0")
```

### 변수의 범위: 전역, 지역

- global 변수 : 함수 밖에 있는 변수. 어디서든 갖다 쓸 수 있음. 
- local 변수 : 함수 내에서만 쓰이는 변수. 밖에서는 못 쓰임. 

```python
global_var = 100 # 전역 변수 : 함수 밖에서 만든 변수. 어디서든 쓸 수 있다. 

def my_func():
    local_var = 20 # 지역 변수 : 함수 안에서 만든 변수. 함수가 끝나면 사라진다. (밖에서 못 쓴다.)
    print(global_var) # 읽기 가능

my_func()
# print(local_var) # 에러! (함수 밖에서는 지역 변수가 안 보임) -- IndentationError: unexpected indent
```

```python
# 함수 안에서 전역 변수를 수정하려면 global 키워드가 필요하지만, 남용하면 코드가 꼬일 수 있어 주의해야 합니다.
# 헷갈려서 되도록 안 쓰는게 좋음. 

count = 0   # 전역 변수

def increase():
    global count
    count = count + 1

increase()
print(count)
```

# 8) 클래스

### 클래스와 객체

- 객체(object, instance) : 변수(데이터, 속성)와 이 데이터를 다루는 기능(함수, 메서드)을 묶어놓은 것 
- 클래스 : 객체를 만들기 위한 서례도. 빈껍데기. 실제 실물은 객체인 것. 데이터와 데이터가 움직이는 함수와 함께 들어있는 것. 

```python
class FishBread:    
    # 생성자: 붕어빵이 태어날 때 실행됨
    # 생성자 메서드 : 만들어야 클래스가 나옴. 클래스 만들 때 def 생성자 메서드와 다른 메서드를 한 개 이상 만듦.
    # __는 특수 메서드. 
    def __init__(self, f, p):           # 언더바를 두개 붙이면 초기 생성 클래스를 만들겠따는 선언.
        self.flavor = f  # 맛 (속성)     - 인스턴스 변수     # self : 객체 자신 
        self.price = p   # 가격 (속성)   - 인스턴스 변수     # f, p 는 매개변수 값 
        # self는 객체로 메서드를 통해 만들어진 실제 값이 들어 있는 결과물 
    # 메서드(클래스 내부의 함수): 붕어빵의 행동
    def show_info(self):  # 함수 형태는 일반 메서드로, 클래스 안에 있는 함수가 메서드인 것. 
        # 이 메서드는 객체뒤에 . 을 찍고 일반 함수 메서드를 붙여서 실행한다. 
        # 여기서 self는 호출한 녀석(bread1 또는 bread2)이 됩니다.
        print(f"{self.flavor} 붕어빵은 {self.price}원입니다.")

# 객체 생성 (틀로 빵 찍어내기)
bread1 = FishBread("팥", 1000)      # self가 bread1이 되어 '팥'을 저장
bread2 = FishBread("슈크림", 1200)  # self가 bread2가 되어 '슈크림'을 저장
bread1.show_info() # bread1의 정보 출력
bread2.show_info() # bread2의 정보 출력
```

### 클래스 변수 vs 인스턴스 변수

인스턴스 변수는 def 안에 있는 self.name 이런 거. 메서드(클래스의 함수) 안에 있는 것이 인스턴스 변수 
count와 같이 def 밖에 있어 클래스 내에서 글로벌 변수 역할을 하는 것이 클래스 변수. 여러 메서드에 적용된다. 

```python
class Monster:
    count = 0  # [클래스 변수] 모든 몬스터가 공유 (총 몬스터 수)

    def __init__(self, name):
        self.name = name  # [인스턴스 변수] 몬스터마다 다름 (내 이름)
        Monster.count += 1 # 몬스터가 태어날 때마다 전체 카운트 증가

m1 = Monster("슬라임")
print(f"현재 몬스터 수: {Monster.count}마리") # 1마리

m2 = Monster("골렘")
print(f"현재 몬스터 수: {Monster.count}마리") # 2마리 (m1에게도 영향을 줌)
```

### 상속

```python
# FishBread의 기능을 물려받음
# 상속한다. 
class PremiumFishBread(FishBread):  # FishBread 클래스의 내용을 통으로 다 가져온다. 
    def discount(self):             # 이렇게 추가적인 정보만 적으면 된다. 
        self.price -= 100 # 100원 할인 기능 추가

p_bread = PremiumFishBread("황금", 2000)    # FishBread의 내용이 그대로 적용된다. 
p_bread.discount()
p_bread.show_info() # 황금 붕어빵은 1900원입니다.
```

### 전기차동차 생산

```python
class Car:
    def __init__(self, model):
        self.model = model
        self.speed = 0

    def accelerate(self, value):
        self.speed += value
        print(f"{self.model} 가속! 현재 속도: {self.speed}km/h")

    def brake(self, value):
        self.speed -= value
        if self.speed < 0:
            self.speed = 0
        print(f"{self.model} 감속! 현재 속도: {self.speed}km/h")
my_car = Car("테슬라")
my_car.accelerate(50)
my_car.brake(20)
```
